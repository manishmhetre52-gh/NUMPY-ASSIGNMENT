#### 27. Consider an integer vector Z, which of these expressions are legal? (★☆☆)
```python
Z**Z
2 << Z >> 2
Z <- Z
1j*Z
Z/1/1
Z<Z>Z
```
`No hints provided...`

````python
z = np.array([1,2,3,4,5])
print(z**z) # [i**i for i in z]
print(2<<z>>2) # element wise bit-shift
print(z<-z) # this checks z is smaller than negative of z ([-1 -2 -3 -4 -5])
print(1j*z) # multiplied by the imaginary number 1j
print(z/1/1) # this means (z/1)/1 => true division, thats why the float o/p
print(z<z>z) # z<z>z means ((z<z)and(z>z)), which is not possible, that's why the error
````

#### 48. Print the minimum and maximum representable values for each numpy scalar type (★★☆)
`hint: np.iinfo, np.finfo, eps`

````python
ints = [np.int8, np.int16, np.int32, np.int64,np.uint8, np.uint16, np.uint32, np.uint64]
for dtype in ints:
    info = np.iinfo(dtype)
    print(dtype.__name__,info.min,info.max)
flts = [np.float16, np.float32, np.float64]
for fdtype in flts:
    info = np.finfo(fdtype)
    print(fdtype.__name__,info.min,info.max)
````
